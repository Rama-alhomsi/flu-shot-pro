```python
df_clean = df.copy()
```

```python
unemployed_mask = df_clean['employment_status'].isin(['Not in Labor Force', 'Unemployed'])
df_clean.loc[unemployed_mask & df_clean['employment_industry'].isnull(), 'employment_industry'] = 'Not Applicable'
df_clean.loc[unemployed_mask & df_clean['employment_occupation'].isnull(), 'employment_occupation'] = 'Not Applicable'
```

```python
high_income_emp = (df_clean['income_poverty'] == '> $75,000') & (df_clean['employment_status'] == 'Employed')
df_clean.loc[high_income_emp & df_clean['health_insurance'].isnull(), 'health_insurance'] = 1
```

```python
df_clean['household_adults'] = df_clean.groupby(['marital_status', 'income_poverty'])['household_adults'].transform(
    lambda x: x.fillna(x.median())
).fillna(df_clean['household_adults'].median())
```

```python
df_clean['household_children'] = df_clean.groupby(['marital_status', 'income_poverty'])['household_children'].transform(
    lambda x: x.fillna(x.median())
).fillna(df_clean['household_children'].median())
```

```python
cat_cols = df_clean.select_dtypes(include=['object']).columns
df_clean[cat_cols] = df_clean[cat_cols].fillna('Unknown')
```
